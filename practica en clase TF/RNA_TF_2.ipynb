```python
import tensorflow as tf
import numpy as np
```

```python
x=np.arange(1,6).reshape(5,1)
#x
```

```python
y=np.array([x*2]).reshape(5)
#y
```

```python
model= tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(1, activation='relu'),
    tf.keras.layers.Dense(1, activation='linear')
])
```

2,5, 1


```python
model.compile(optimizer='sgd', loss='mse')
```

```python
model.fit(x,y,epochs=10)
```

```python
predicciones=model.predict(np.array([[6],[7]]))
predicciones
```

```python
x = np.array([
 [0, 1, 2], [1, 2, 3], [2, 3, 4], [3, 4, 5],
 [4, 5, 6], [5, 6, 7], [6, 7, 8], [7, 8, 9],
 [8, 9, 10], [9, 10, 11], [10, 11, 12], [11, 12, 13],
 [12, 13, 14], [13, 14, 15], [14, 15, 16], [15, 16, 17],
 [16, 17, 18], [17, 18, 19], [18, 19, 20], [19, 20, 21],
 [20, 21, 22], [21, 22, 23], [22, 23, 24], [23, 24, 25],
 [24, 25, 26], [25, 26, 27], [26, 27, 28], [27, 28, 29],
 [28, 29, 30], [29, 30, 31], [30, 31, 32], [31, 32, 33]
], dtype=np.float32)

y = np.array([
 3, 6, 9, 12, 15, 18, 21, 24,
 27, 30, 33, 36, 39, 42, 45, 48,
 51, 54, 57, 60, 63, 66, 69, 72,
 75, 78, 81, 84, 87, 90, 93, 96
], dtype=np.float32)

model = tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(3,)),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1, activation='linear')
])

model.compile(optimizer='adam', loss='mse')
model.fit(x, y, epochs=150, verbose=0)

# Predicción coherente con el patrón
print(model.predict(np.array([[5, 6, 7]], dtype=np.float32)))
```
