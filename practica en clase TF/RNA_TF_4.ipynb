```python
import tensorflow as tf
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_wine
```

```python
vino = load_wine()
x = vino.data   #x debe ser 2d
y = vino.target #y debe ser 1d
```

```python
#---Análisis exploratorio---
print('Atributos del datase', vino.feature_names)
print('Entradas para mi RNA', len(vino.feature_names))
print('Cantidad de muestras en total',len(y))
print('Etiquetas', vino.target_names)
print('Salidas para mi RNA', len(vino.target_names))
print('Ejemplo de muestra: ', x[149])
print('Ejemplo de etiqueta', y[149])
print('Ejemplo de nombre de etiqueta', vino.target_names[y[149]])
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.2, random_state=42)
print('Cantidad de muestras para entrenamiento', len(y_train))
print('Cantidad de muestras para pruebas del modelo', len(y_test))
```

```python
model= tf.keras.models.Sequential([
    tf.keras.layers.Dense(32,activation='relu', input_shape=(13,)),
    tf.keras.layers.Dense(66, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')
model.fit(x_train, y_train, epochs=100)
```

```python
print('Predicción individual')
muestra= x_test[0].reshape(1,13)
print('muestra seleccionada', muestra)
print('etiqueta real de la muestra', y_test[0])
prediccion= model.predict(muestra)
print('predicción del modelo (Probabilidades)', prediccion)
print('predicción del modelo (Afinada)', prediccion.argmax(1))
```
