```python
import pandas as pd
```

```python
boleta= pd.DataFrame({
    "idalumno":[1,2,3,4,5],
    "nombre":['SUSSIE','PEPE','ALBERTO','MATILDE','LEONY'],
    "u1":[12,15,14,16,11],
    "u2":[20,15,16,17,5],
    "u3":[13,13,14,13,5]
})
boleta
```

```python
boleta["promedio"] = (boleta["u1"] + boleta["u2"] + boleta["u3"]) / 3
boleta
```

```python
boleta["promedio"] = boleta["promedio"].round(2)
boleta
```

```python
def estado_alumno(promedio):
    if promedio >= 10.5:
        return "APROBADO"
    else:
        return "DESAPROBADO"

boleta["estado"] = boleta["promedio"].apply(estado_alumno)
boleta
```

```python
boleta["estado"].value_counts()
```

```python
boleta.loc[boleta["promedio"].idxmax()]  # el de mejor promedio
boleta.loc[boleta["promedio"].idxmin()]  # el de peor promedio
```

```python
boleta.sort_values("promedio", ascending=False)
```
