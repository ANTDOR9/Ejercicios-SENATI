```python
import pandas as pd
```

```python
productos = pd.DataFrame({
    "id":[1,2,3],
    "nombre":["leche gloria" , "cocacola 9L" ,"SUBLIME"],
    "monto": [5.5, 6.5, 2.0]
    })
productos
```

```python
productos.sort_values("monto", ascending=False)
```

```python
productos["igv"] = productos ["monto"]*0.18
productos
```

```python
productos["igv"] = productos["igv"].round(2)
```

```python
print(productos)
```

```python
def clasificar(monto):
    if monto > 5:
        return "caro"
    else:
        return "barato"

productos["categoria"] = productos["monto"].apply(clasificar)
productos
```
