```python
import nltk
```

```python
from nltk.tokenize import word_tokenize, TweetTokenizer
```

```python
nltk.download("puntk")  #trabaje correctamente
texto = "hola a todos ¿como estan?, hoy atenderemos NLP"
```

```python
ventas=[2500,500,1200,200,800]
for v in ventas:
    if v > 1000:
        print(v)

print([v for v in ventas if v > 1000]) 
```

```python
print("---tokenizando con word_tokenize----")
tokens = word_tokenize(texto.lower())
print(tokens)
```

```python
print("---tokenizando con word_tokenize----")
tokenizer = TweetTokenizer(texto.lower())
tokens = tokenizer.tokenize(texto.lower())
print(tokens)
```
