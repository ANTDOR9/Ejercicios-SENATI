```python
import spacy
```

```python
spacy.cli.download("es_core_news_sm")
```

```python
nlp=spacy.load("es_core_news_sm")
```

```python
texto="hola a todos ¿Cómo están?, hoy aprenderemos NLP"
tokens=nlp(texto.lower())
print("-----resultados token-----")
print([t for t in tokens])
```

```python
nlp=spacy.load("es_core_news_sm")
texto="El instructor Pepe dictará un curso en SENATI"
tokens=nlp(texto)
print("----resultados token----")
print([t for t in tokens])

print("---lematizacion---")
for t in tokens:
    print(t.lemma_)

print("---ETIQUETADO GRAMTICAL (POS TAGGING)---")
for t in tokens:
    print(t.text, t.tag_)


print("---Reconomineto de indentificadores---")
for t in tokens.ents:
    print(t)
```

```python
#tarea---------------------------------------
print("---Reconocimiento de lista---")
for t in tokens:
    print(([t.text, t.tag_, t.lemma_]))
```

```python
print("---ETIQUETADO GRAMTICAL (POS TAGGING II)---")
print([(t.text, t.tag_) for t in tokens])   
```
