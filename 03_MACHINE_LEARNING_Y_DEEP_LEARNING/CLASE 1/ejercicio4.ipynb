```python
import numpy as np
from sklearn.decomposition import PCA
```

```python
x = np.array([[1,1], [2,2], [3,3], [4,4]])
x
```

```python
pca= PCA(n_components=1)
pca_reduced = pca.fit_transform(x)
pca_reduced
```

```python
x_reformed= pca.inverse_transform(pca_reduced)
x_reformed
```
