```python

import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris


import matplotlib.pyplot as plt
```

```python

x, _ = load_iris(return_X_y=True)
model = KMeans(n_clusters=3, random_state=42)
```

```python
model.fit(x)
```

```python

centros = model.cluster_centers_
etiquetas = model.labels_
```

```python

print("Etiquetas", model.labels_)
print("Centroides", model.cluster_centers_)
     
```

```python

plt.scatter(x[:, 0], x[:, 1], c=model.labels_, cmap='viridis', label='Flores')
plt.scatter(centros[:, 0], centros[:, 1], marker='X', s=100, c='red', label='Centroides')
```

```python
iris= load_iris()
x =iris.data
y= iris.target
```

```python

model = KMeans(n_clusters=3, random_state=42)
model.fit_predict(x) #Entrenamos el modelo (utilizando las 4 caracteristicas)
```

```python

etiquetas = model.labels_
centros = model.cluster_centers_
```

```python

print("Etiquetas", model.labels_)
print("Centroides", model.cluster_centers_)
```

```python

plt.scatter(x[:, 0], x[:, 1], c=model.labels_, cmap='viridis', label='Flores')
plt.scatter(centros[:, 0], centros[:, 1], marker='X', s=100, c='red', label='Centroides')
```

```python


plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
plt.title("CLUSTERING POR KMEANS")
plt.scatter(x[:,0],x[:,1], c= etiquetas)
plt.subplot(1,2,2)
plt.title("DATA REAL")
plt.scatter(x[:,0],x[:,1], c=y)
```

```python

print(etiquetas)
print(y)
print("cantidad de clusters", len(set(etiquetas)))
print("muestras por cluster", np.bincount(etiquetas))
print("muestras por flor (real)", np.bincount(y))
```
