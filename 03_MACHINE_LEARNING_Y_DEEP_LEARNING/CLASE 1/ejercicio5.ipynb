```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler  #NORMALIZAR
from sklearn.cluster import KMeans  #CLUSTERIZAR
from sklearn.decomposition import PCA  #REDUCIR LA DIMENSIONALIDAD
from sklearn.metrics import silhouette_samples, adjusted_rand_score, normalized_mutual_info_score  #METRICAS PAR AEL MODELO
```

```python
iris =load_iris()
x = iris.data [:,[2,3]]  #INGRESO DE LAS VARIABLE MEJOR RANKEADAS...DESPUESD EL []
y = iris.target  #PARA EVALUAR AL MODELO
especies = iris.target_names
```

```python
x_scaled= StandardScaler() .fit_transform(x)  #NORMALIZAR LOS DATOS
x_2d= PCA(n_components=2).fit_transform(x_scaled)  #REDUCIRLOS LA DIMENISONALIDA A 2D
#x_2d
```

```python
model= KMeans(n_clusters=3, n_init=10, random_state=42)
model.fit_predict(x_scaled)
```

```python
plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
plt.title("clustering")
plt.scatter(x[:,0], x[:,1], c=model.labels_)
plt.subplot(1,2,2)
plt.scatter(x[:,0], x[:,1], c=y)
plt.title("datos reales")
```
