# 🌸 Práctica 02 — K-Means con datos reales (Iris)

**Objetivo:** aplicar K-Means a un dataset real y **comparar** los grupos que encuentra la máquina con las etiquetas verdaderas.

> 💡 Aunque Iris tiene etiquetas (`y`), **no se las daremos al modelo**. Solo las usaremos al final para ver *qué tan bien adivinó*.

Repaso de la práctica 01: `reshape`, `KMeans(n_clusters=k)`, `.fit`, `.labels_`, `.cluster_centers_`.

## Bloque 1 — Herramientas y datos

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris

iris = load_iris()
x = iris.data       # medidas de las flores (lo que el modelo VE)
y = iris.target     # especie real: 0, 1, 2 (el modelo NO lo ve)
```

Ya viene como tabla, así que **no hace falta `reshape`**. Miremos qué tenemos:

```python
print("Forma de x:", x.shape, "→ 150 flores × 4 características")
print("Características:", iris.feature_names)
print("Especies:", iris.target_names)
print("Primeras 5 filas:\n", x[:5])
```

❓ **Pregunta:** ¿cuántas flores hay? ¿cuántas medidas tiene cada una?

## Bloque 2 — Entrenar con k = 3
Sabemos que hay 3 especies, así que pedimos 3 grupos.

```python
model = KMeans(n_clusters=3, random_state=42, n_init=10)
model.fit(x)

etiquetas = model.labels_
centros = model.cluster_centers_
print("Primeras 15 etiquetas:", etiquetas[:15])
print("Forma de los centroides:", centros.shape)
```

Los centroides tienen forma **(3, 4)**: 3 grupos, y cada centro es un punto con 4 coordenadas (una por característica).

## Bloque 3 — Graficar usando 2 de las 4 características
No podemos dibujar 4 dimensiones, así que elegimos las columnas 0 y 1 (largo y ancho del sépalo).

```python
plt.scatter(x[:, 0], x[:, 1], c=etiquetas, s=40)
plt.scatter(centros[:, 0], centros[:, 1], marker="X", s=200, c="red", label="Centroides")
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.title("K-Means sobre Iris")
plt.legend()
plt.show()
```

## Bloque 4 — ¿Y la realidad? Lado a lado
Dibujamos los grupos del modelo junto a las especies reales.

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.title("Clustering por K-Means")
plt.scatter(x[:, 0], x[:, 1], c=etiquetas)

plt.subplot(1, 2, 2)
plt.title("Datos reales")
plt.scatter(x[:, 0], x[:, 1], c=y)

plt.show()
```

> 🧠 `plt.subplot(1, 2, 1)` = *1 fila, 2 columnas, dibuja en la posición 1*.

❓ Los **colores** pueden no coincidir entre ambos gráficos (el grupo "0" del modelo no tiene por qué ser la especie 0). Lo importante es si los **grupos** se parecen.

## Bloque 5 — Comparar con números

```python
print("Muestras por grupo (modelo):", np.bincount(etiquetas))
print("Muestras por especie (real):  ", np.bincount(y))
```

`np.bincount` cuenta cuántas veces aparece cada valor (0, 1, 2).

Esto es una comparación aproximada. Para ver **qué especie cayó en qué grupo** usamos una tabla cruzada:

```python
import pandas as pd

tabla = pd.crosstab(y, etiquetas, rownames=["Especie real"], colnames=["Grupo K-Means"])
tabla
```

Cada fila es una especie real, cada columna un grupo del modelo.

- Una especie concentrada en **un solo grupo** → el modelo la separó bien.
- Una especie repartida en **varios grupos** → el modelo se confundió.
