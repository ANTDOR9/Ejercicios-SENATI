```python
#obtener las etiquetas y asegurarse que se encuentran en 1D (y)
#mediante PCA reducir X a 1D
#IMPLEMENTAR LA RNA DE CLASFICACIÓN BINARIA
```

```python
pclass = titanic['Pclass'].to_numpy().reshape(-1, 1)                 # (891, 1)
sex = LabelBinarizer().fit_transform(titanic['Sex'])    # (891, 1)
age = titanic['Age'].fillna(titanic['Age'].median()).to_numpy().reshape(-1, 1)
age = StandardScaler().fit_transform(age) 
```

```python
X = np.hstack([pclass, sex, age])  
```

```python
y = titanic['Survived'].to_numpy().ravel()        # (891,)
print(X.shape, y.shape)
```

```python
X = StandardScaler().fit_transform(X)    # PCA funciona mejor con datos en la misma escala
pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)                                         # (891, 1)
print(X_pca.shape, pca.explained_variance_ratio_)
```

```python
# --- 3) RNA de clasificación binaria ---
X_train, X_test, y_train, y_test = train_test_split(
    X_pca, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(4, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')   # salida binaria
])
```

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

history = model.fit(X_train, y_train, epochs=50, batch_size=32,
                    validation_split=0.2, verbose=1)
```

```python
y_pred = (model.predict(X_test) > 0.5).astype(int).ravel()
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
