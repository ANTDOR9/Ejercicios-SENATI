# 02 · Regresión con TensorFlow/Keras## ObjetivoEntender cómo una red neuronal puede predecir un **valor numérico continuo**, en vez de una categoría.## Teoría: clasificación vs. regresión| | Clasificación | Regresión ||---|---|---|| ¿Qué predice? | Una categoría (0/1, spam/no spam) | Un número continuo (precio, temperatura) || Activación de salida típica | sigmoid (2 clases) o softmax (varias) | linear (sin restricción de rango) || Función de pérdida típica | binary_crossentropy / categorical_crossentropy | mse (error cuadrático medio) |En este notebook vamos a enseñarle a la red la relación **Y = 3X** (la tabla del 3), para ver si logra "descubrir" esa fórmula solo mirando ejemplos.

## Paso 1: Importar libreríasIgual que en el notebook anterior: TensorFlow para el modelo, NumPy para los datos.

```python
import tensorflow as tfimport numpy as np
```

## Paso 2: Preparar los datosUsamos la tabla del 3: si X=1, Y=3; si X=2, Y=6; y así sucesivamente. La idea es que el modelo aprenda la relación **Y = 3·X** solo con estos ejemplos, sin que le digamos la fórmula directamente.

```python
X = np.array([[1], [2], [3], [4], [5], [6]])Y = np.array([3, 6, 9, 12, 15, 18])print("X:", X.flatten())print("Y:", Y)
```

## Paso 3: Construir la arquitecturaDiferencias clave respecto al notebook de clasificación:- La capa oculta ahora tiene **4 neuronas** con ReLU (más capacidad que el ejemplo anterior, aunque el problema es simple).- La capa de salida usa activación **"linear"**: significa que NO se aplica ninguna función de activación extra, la neurona entrega directamente el valor `Z = ΣXᵢWᵢ + b`. Esto es necesario en regresión porque el valor a predecir puede ser cualquier número, no solo algo entre 0 y 1.

```python
modelo = tf.keras.Sequential([    tf.keras.layers.Dense(4, activation="relu", input_shape=(1,)),    tf.keras.layers.Dense(1, activation="linear")  # regresión: sin restricción de rango])modelo.summary()
```

## Paso 4: Compilar- **optimizer="sgd"** (Stochastic Gradient Descent): un optimizador más simple y "clásico" que Adam, bueno para ver el aprendizaje de forma más gradual.- **loss="mse"** (Mean Squared Error / error cuadrático medio): mide el promedio de (predicción − valor real)² . Es la pérdida estándar para regresión, porque penaliza más los errores grandes.

```python
modelo.compile(optimizer="sgd", loss="mse")
```

## Paso 5: EntrenarUsamos más épocas (80) porque SGD suele necesitar más pasadas que Adam para converger.

```python
historial = modelo.fit(X, Y, epochs=80, verbose=0)print("Pérdida (mse) final:", historial.history["loss"][-1])
```

## Paso 6: PredecirProbamos con X=10. Si el modelo aprendió bien la relación Y=3X, la predicción debería acercarse a 30.

```python
prediccion = modelo.predict(np.array([[10]]))print("Predicción para X=10 (esperado ≈30):", prediccion)
```

## Conclusión del notebook- La diferencia técnica clave entre clasificar y hacer regresión está en la **capa de salida**: sigmoid/softmax para clasificar, linear para regresión.- También cambia la **función de pérdida**: crossentropy para clasificar, mse para regresión.- El resto del flujo (armar el modelo, compilar, entrenar, predecir) es idéntico.
