```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_lfw_people
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.svm import SVC #máquinas de soporte de vectores
```

ESTE ES UNA PRACTICA DEL PROFESOR. TAREA:
MEJORAR EL CODIGO MOSTRADO

```python
faces= fetch_lfw_people(min_faces_per_person=70, resize=0.4)
```

```python
x= faces.data
y= faces.target
nombres= faces.target_names
h,w = faces.images.shape[1:]
print((h,w))
print(len(nombres))
print(nombres)
print(x.shape[0])
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
model= SVC(kernel='linear', class_weight='balanced') #CONFIGURAR OTROS PARÁMETROS
model.fit(x_train, y_train)
```

```python
y_pred= model.predict(x_test)
print('Predicciones')
print(y_pred)
print('Etiquetas reales')
print(y_test)
```

```python
print('métricas')
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred, target_names=nombres))
```

```python
for i in range(12):
  plt.subplot(3,4, i+1)
  plt.imshow(x_test[i].reshape(50,37), cmap='gray')
  plt.title(str(faces.target_names[y_pred[i]])+'\n'+str(faces.target_names[y_test[i]]))
```
