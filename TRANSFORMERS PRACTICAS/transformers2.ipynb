```python
import torch
from transformers import pipeline

print("Torch disponible:", torch.__version__)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

modelo_nombre = "Helsinki-NLP/opus-mt-es-en"
tokenizer = AutoTokenizer.from_pretrained(modelo_nombre)
modelo = AutoModelForSeq2SeqLM.from_pretrained(modelo_nombre)
```

```python
texto = "Hola, soy Anthony, vivo en Arequipa. ¿Donde vives tú?"
entradas = tokenizer(texto, return_tensors="pt")
salida = modelo.generate(**entradas)
traduccion = tokenizer.decode(salida[0], skip_special_tokens=True)
print(traduccion)
```
