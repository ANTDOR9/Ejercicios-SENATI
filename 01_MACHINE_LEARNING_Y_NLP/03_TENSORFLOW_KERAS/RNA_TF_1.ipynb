```python
import tensorflow as tf
import numpy as np
```

```python
x=np.linspace(-10,10,100).reshape(100,1)
#x #quitar el # para ejecutar y ver como quedan los datos x
```

```python
y=np.where(x>0,1,0).reshape(100)
#y #quitar el # para ejecutar y ver como quedan las etiquetas y
```

```python
model= tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(15, activation='relu'),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy')
```

```python
model.fit(x,y,epochs=50)
```

```python
predicciones=model.predict(np.array([[-1220],[24440],[-10],[10],[-15],[15],[-25],[20],[-0]]))
print('Salida real de la RNA\n',predicciones)
print('Salida ajustada (Redondeada)\n', predicciones.round())
```
