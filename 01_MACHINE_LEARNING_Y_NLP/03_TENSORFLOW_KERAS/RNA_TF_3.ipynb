```python
import tensorflow as tf
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
x = iris.data   #x debe ser 2d
y = iris.target #y debe ser 1d
```

```python
#---Análisis exploratorio---
print("Dato",x[0])
print("Etiqueta",y[0])
print("Nombre Etiqueta",iris.target_names[y[149]])
print("Caràcteristicas del dataset", iris.feature_names)
print("cant. de caracterìsticas (Para input layer:)", len(iris.feature_names))
print("etiquetas:", len(iris.target_names))
print(" salidas par mi RNA----cant. de etiquetas (Para output layer):", len(iris.target_names))
print("ejemplo de muestra", x[149])
print("ejemplo de etiqueta", y[149])
print("ejemplo de nombre de etiqueta", iris.target_names[y[149]])
```

```python
#---SPLIT DE LA DATA PARA ENTRENAMIENTO 80% Y PRUEBA 20%---
x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.2, random_state=42)
print('Cantidad de muestras para entrenamiento', len(y_train))
print('Cantidad de muestras para pruebas del modelo', len(y_test))
```

```python
#---CONSTRUCCIÓN DE NUESTRA RNA
model= tf.keras.models.Sequential([
    tf.keras.layers.Dense(10,activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

32,66

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')
model.fit(x_train, y_train, epochs=100)
```

```python
print('Predicción individual')
muestra= x_test[0].reshape(1,4)
print('muestra seleccionada', muestra)
print('etiqueta real de la muestra', y_test[0])
prediccion= model.predict(muestra)
print('predicción del modelo (Probabilidades)', prediccion)
print('predicción del modelo (Afinada)', prediccion.argmax(1))
```
