# 03 · Plantillas de referencia: ¿qué arquitectura uso?

## Objetivo
Este notebook no entrena ningún modelo. Es una **hoja de referencia rápida** para decidir, según el tipo de problema, qué activación de salida y qué función de pérdida usar en Keras — muy útil para no tener que memorizar todo antes de la exposición o de un examen. (acutlaizado)

## Teoría: la tabla de decisión

| Tipo de problema | Activación de salida | Neuronas de salida | Función de pérdida |
|---|---|---|---|
| Regresión (predecir un número) | `linear` | 1 | `mse` |
| Clasificación binaria (2 clases) | `sigmoid` | 1 | `binary_crossentropy` |
| Clasificación multiclase (3+ clases) | `softmax` | tantas como clases | `sparse_categorical_crossentropy` (etiquetas como enteros) o `categorical_crossentropy` (etiquetas one-hot) |

A continuación, una plantilla de código por cada caso. Ninguna se ejecuta con datos reales aquí — son "moldes" para copiar y adaptar.


## Plantilla A — Regresión

Salida `linear`, pérdida `mse`. Úsala cuando el resultado que buscas es un número (precio, temperatura, cantidad).


```python
# --- PLANTILLA: REGRESIÓN ---
import tensorflow as tf

modelo_regresion = tf.keras.models.Sequential([
    tf.keras.layers.Dense(4, activation="relu", input_shape=(1,)),
    tf.keras.layers.Dense(1, activation="linear")
])
modelo_regresion.compile(optimizer="sgd", loss="mse")
```

## Plantilla B — Clasificación binaria

Salida `sigmoid` (1 neurona), pérdida `binary_crossentropy`. Úsala cuando la respuesta es una de dos categorías.


```python
# --- PLANTILLA: CLASIFICACIÓN BINARIA ---
import tensorflow as tf

modelo_binario = tf.keras.models.Sequential([
    tf.keras.layers.Dense(8, activation="relu", input_shape=(1,)),
    tf.keras.layers.Dense(1, activation="sigmoid")
])
modelo_binario.compile(optimizer="adam", loss="binary_crossentropy")
```

## Plantilla C — Clasificación multiclase

Salida `softmax`, con **una neurona por clase posible**, pérdida `sparse_categorical_crossentropy` (cuando las etiquetas son números enteros como 0, 1, 2...).


```python
# --- PLANTILLA: CLASIFICACIÓN MULTICLASE (ej. 3 clases) ---
import tensorflow as tf

modelo_multiclase = tf.keras.models.Sequential([
    tf.keras.layers.Dense(10, activation="relu", input_shape=(4,)),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(3, activation="softmax")
])
modelo_multiclase.compile(optimizer="adam", loss="sparse_categorical_crossentropy")
```

## Conclusión del notebook

La pregunta que siempre hay que hacerse antes de armar la última capa es: **¿cuántas respuestas posibles tiene mi problema, y son categorías o un número?**

- ¿Un número cualquiera? → Plantilla A (regresión).
- ¿Solo 2 opciones? → Plantilla B (binaria).
- ¿3 o más opciones? → Plantilla C (multiclase).
