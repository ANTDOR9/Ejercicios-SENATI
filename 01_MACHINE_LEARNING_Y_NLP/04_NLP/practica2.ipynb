```python
from nltk.tokenize import word_tokenize, TweetTokenizer  #llamando a los modulos
```

```python
from nltk.corpus import stopwords  #coleccion de documentos para entrenar
```

```python
print(stopwords.words("spanish")) #copus de palabras que no aportan sentido gramatical a las oraciones o textos
```

```python
stop_words = stopwords.words("spanish")
texto="hola a todos ¿cómo están?, hoy atenderemos NLP"
tokens= word_tokenize (texto.lower())
texto_limpio= [t for t in tokens if t.isalpha () and not t in stop_words]
print(texto_limpio)
```

```python
stop_words = stopwords.words("spanish")
texto2="hola a todos ¿cómo están?, hoy atenderemos NLP"

print("---tecnica con tweet_tokenize limpiando----")

tokenizer = TweetTokenizer(texto2.lower())
tokens = tokenizer.tokenize(texto2.lower())


texto_limpio2= [t for t in tokens if t.isalpha () and not t in stop_words]

print(texto_limpio2)
```
