```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

ventas = pd.DataFrame({
    'vendor': [
        'fabi', 'fabi','gustabo', 'gustabo','antoni del cine', 'antoni del cine','antoni viejo',
        'antoni viejo','matias', 'matias'
        ],'ventas': [200, 500,400, 600,300, 500,400, 800, 700, 900
  ]})
```

```python
# --- Funciones de varias filas ---
# Venta máxima
venta_max = ventas['ventas'].max()
print("Venta máxima:", venta_max)
# Conteo de ventas
conteo_ventas = ventas['ventas'].count()
print("Conteo de ventas:", conteo_ventas)
# Media de ventas
media_ventas = ventas['ventas'].mean()
print("Media de ventas:", media_ventas)
# Suma de ventas
suma_ventas = ventas['ventas'].sum()
print("Suma de ventas:", suma_ventas)
```

```python
# Trabajando con Gustabo
# Todo de Gustabo
print(ventas[ventas['vendor'] == 'gustabo'])
# Suma de las ventas de Gustabo
print(ventas[ventas['vendor'] == 'gustabo']['ventas'].sum())
# Venta máxima de Gustabo
print(ventas[ventas['vendor'] == 'gustabo']['ventas'].max())
```

```python
# Todo de Antoni viejo
print(ventas[ventas['vendor'] == 'antoni viejo'])

# Suma de las ventas de Antoni viejo
print(ventas[ventas['vendor'] == 'antoni viejo']['ventas'].sum())

# Todo de Gustabo
print(ventas[ventas['vendor'] == 'gustabo'])

# Suma de las ventas de Gustabo
print(ventas[ventas['vendor'] == 'gustabo']['ventas'].sum())

# --- Trabajando con Antoni viejo y Gustabo ---

# Venta máxima de Antoni viejo
print("Venta máxima de Antoni viejo:",
      ventas[ventas['vendor'] == 'antoni viejo']['ventas'].max())

# Promedio de todas las ventas de Antoni viejo
print("Promedio de Antoni viejo:",
      ventas[ventas['vendor'] == 'antoni viejo']['ventas'].mean())

# Venta máxima de Gustabo
print("Venta máxima de Gustabo:",
      ventas[ventas['vendor'] == 'gustabo']['ventas'].max())

# Promedio de todas las ventas de Gustabo
print("Promedio de Gustabo:",
      ventas[ventas['vendor'] == 'gustabo']['ventas'].mean())
```

```python
# --- TRABAJANDO CON TODOS LOS VENDEDORES ---

# SUMA TOTAL POR VENDEDOR
suma_por_vendedor = ventas.groupby('vendor')['ventas'].sum()
print(suma_por_vendedor)

# PROMEDIO DE VENTAS POR VENDEDOR
promedio_por_vendedor = ventas.groupby('vendor')['ventas'].mean()
print(promedio_por_vendedor)
```

```python
# --- COMBINANDO GRÁFICOS ---

print('Suma por vendedor:')
print(suma_por_vendedor.index)
print(suma_por_vendedor.values)

print('Promedio por vendedor:')
print(promedio_por_vendedor.index)
print(promedio_por_vendedor.values)

plt.bar(suma_por_vendedor.index, suma_por_vendedor.values, label='Suma')
plt.bar(promedio_por_vendedor.index, promedio_por_vendedor.values, label='Promedio')

plt.title('Suma y promedio por vendedor')
plt.xlabel('Vendedor')
plt.ylabel('Ventas')
plt.legend()
plt.xticks(rotation=45)

plt.show()
```
