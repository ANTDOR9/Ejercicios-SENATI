```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
ventas= pd.DataFrame({
    'vendedor':['Sussie','Pepe','Sussie','Sussie','Pepe','Alberto', 'Alberto'],
    'venta':[1000, 500, 1200, 500, 1000, 2000, 1000]
})
ventas
```

```python
#---Funciones de varias filas---
#venta máxima (max)
venta_max= ventas['venta'].max()
print(venta_max)
#conteo de ventas (count)

#media de ventas (mean)
media_ventas= ventas.venta.mean()
print(media_ventas)
#suma de las ventas (sum)
```

```python
#---Trabajando con Sussie---
#ventas.groupby('vendedor')['venta'].sum()
#Todo de Sussie
print(ventas[ventas['vendedor']=='Sussie'])
#Suma de las ventas de Sussie
print(ventas[ventas['vendedor']=='Sussie']['venta'].sum())
#---Trabajando con Pepe y Sussie---
#Venta máxima de Pepe

#Promedio de todas las ventas de pepe

#Promedio de todas las ventas de sussie
```

```python
#---TRABAJANDO CON TODOS LOS VENDEDORES---
#SUMA TOTAL POR VENDEDOR
suma_por_vendedor= ventas.groupby('vendedor')['venta'].sum()
print(suma_por_vendedor)
#PROMEDIO DE VENTAS POR VENDEDOR
promedio_por_vendedor= ventas.groupby('vendedor')['venta'].mean()
print(promedio_por_vendedor)
```

```python
#---COMBINANDO REPORTES ESTADÍSTICOS EN UN GRÁFICO---
print('Suma por vendedor:')
print(suma_por_vendedor.index)
print(suma_por_vendedor.values)
print('Promedio por vendedor:')
print(promedio_por_vendedor.index)
print(promedio_por_vendedor.values)
plt.bar(suma_por_vendedor.index, suma_por_vendedor.values, label='Suma', alpha=0.5)
plt.bar(promedio_por_vendedor.index, promedio_por_vendedor.values, label='Promedio', alpha=0.5)
plt.legend()
```
