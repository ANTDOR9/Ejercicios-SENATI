```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
familia= pd.DataFrame({
    'idfamilia':[1,2,3],
    'nombre':['Bebidas','Lácteos','Escolar']
})
familia
```

```python
categoria=pd.DataFrame({
    'idcategoria':[1,2,3,4],
    'nombre':['Agua Mineral','Gaseosa','Leche','Queso'],
    'idfamilia':[1,1,2,2]
})
categoria
```

```python
reporte1=categoria.merge(familia, on='idfamilia', how='inner')
reporte1=reporte1.rename(columns={'nombre_x':'categoria','nombre_y':'familia'})
reporte1
```

```python
conteo_por_familia= reporte1.groupby('familia')['categoria'].count()
print
conteo_por_familia
```

```python
#primero estructurar el reporte estadístico
conteo_por_familia= reporte1.groupby('familia')['categoria'].count()
#obtener las categorias o etiquetas
print(conteo_por_familia.index)
#obtener los valores numéricos
print(conteo_por_familia.values)
conteo_por_familia
```

```python
cat= conteo_por_familia.index
val= conteo_por_familia.values
col= ['r','g']
plt.bar(cat, val, color= col)
plt.legend()
```

```python
etiquetas=['Bebidas','Lácteos']
for i, (nombre, valor) in enumerate(zip(cat, val)):
  plt.bar(nombre, valor, color= col[i], label= etiquetas[i])
plt.legend()
```

```python
plt.pie(val, labels= cat, autopct='%1.1f%%')
plt.legend()
```

```python
fig, axes= plt.subplots(1,2) #DIVIDIMOS LA FIGURA PRINCIPAL EN 1 FILA POR 2 COLUMNAS (2 CELDAS)
axes[0].bar(cat, val, color= col) #EN LA PRIMERA: GRAFICO DE BARRAS
axes[0].set_title('Barras')
axes[1].pie(val, labels= cat, autopct='%1.1f%%')#EN LA SEGUNDA GRÁFICO DE PASTEL
axes[1].set_title('Pastel')
```
