# Pauta — Regresión Logística y Regresión Lineal con datos simples

**Objetivo de esta práctica:** ver, con datos inventados y muy pequeños, la diferencia entre los dos modelos lineales de scikit-learn antes de usarlos con datasets reales:

- **Regresión Logística** → predice una **clase** (0 o 1).
- **Regresión Lineal** → predice un **número** continuo.

## 1. Importar librerías

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
```

## 2. Regresión Logística

**Datos:** los números del -10 al 9. La etiqueta es 1 si el número es positivo y 0 si no lo es. `reshape(-1, 1)` convierte la lista en una columna, que es la forma que pide scikit-learn.

```python
x = np.arange(-10, 10).reshape(-1, 1)
x
```

```python
y = np.where(x > 0, 1, 0).reshape(-1)       # 1 si x > 0, 0 en otro caso
y
```

```python
model = LogisticRegression().fit(x, y)
```

Se predicen números que el modelo no vio: dos negativos y dos positivos.

```python
model.predict(np.array([[-21], [-30], [11], [15]]))
```

## 3. Regresión Lineal

**Datos:** los números del 1 al 10, y como resultado el doble de cada uno (y = 2x). El modelo debe aprender esa relación y extenderla a valores nuevos.

```python
from sklearn.linear_model import LinearRegression
```

```python
x = np.arange(1, 11).reshape(-1, 1)
x
```

```python
y = np.array(x * 2).reshape(-1)             # y = 2x
y
```

```python
model = LinearRegression().fit(x, y)
model.predict(np.array([[11], [12]]))       # Se espera 22 y 24
```

## 🗒️ Conclusión

Ambos modelos usan el mismo flujo (`fit` → `predict`). La Regresión Logística devolvió clases (0 para negativos, 1 para positivos) y la Lineal devolvió números (22 y 24), confirmando que aprendió la relación y = 2x.
