```python
from sklearn.feature_selection import SelectKBest, f_classif   #calcular cuales son los datos que mantienen mayor eficiencia para el modelo, si la intension es 2 mejores caracterisits es Fregression 
from sklearn.datasets import load_breast_cancer 

```

```python
data= load_breast_cancer()  # Cargamos el conjunto de datos de cáncer de mama
x= data.data  # Obtenemos las características (features)
y= data.target  # Obtenemos las etiquetas (labels)
```

```python
selector = SelectKBest(k=2, score_func=f_classif)  # Creamos un selector de características que seleccionará las 2 mejores características según la prueba ANOVA F
selector.fit_transform(x, y)  # Ajustamos el selector a los datos y transformamos las características
```

```python
indices= selector.get_support(indices=True)  # Obtenemos los índices de las características seleccionadas
print("Indices de las características seleccionadas:", indices)  # Imprimimos los índices de las características seleccionadas
print(data.feature_names[indices])  # Imprimimos los nombres de las características seleccionadas
```
