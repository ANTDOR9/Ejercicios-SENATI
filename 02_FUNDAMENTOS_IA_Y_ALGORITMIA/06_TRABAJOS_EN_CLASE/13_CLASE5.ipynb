```python
import pandas as pd

datos = {

  "experiencia": [1, 2, 3, 5, 7, 10, 4, 6, 8, 2, 9, 3],
  "lenguaje": [
    "python", "java", "csharp", "python",
    "java", "csharp", "java", "python",
    "csharp", "python", "java", "csharp"
  ],

  "rango": [

    "junior", "junior", "intermedio", "intermedio",
    "senior", "senior", "intermedio", "senior",
    "senior", "junior", "senior", "intermedio"

  ],

  "pret_salarial": [
    1500, 1800, 2500, 3000,
    4500, 5000, 2800, 4200,
    4800, 1700, 5500, 2700

  ],

  "contratado": [
    "si", "no", "si", "si",
    "si", "si", "no", "si",
    "si", "no", "si", "no"
  ]

}

df = pd.DataFrame(datos)
print(df)
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder, LabelBinarizer

# TRANSFORMADORES DE DATOS
# ------------------------------------------------------------------------------
# StandardScaler  -> variables NUMÉRICAS (x).
#                    Escala cada columna a media 0 y desviación 1.
#                    Ej.: edad, sueldo, años de experiencia.
#
# OneHotEncoder   -> variables CATEGÓRICAS NOMINALES (x), sin orden.
#                    Crea una columna por categoría con 1 o 0.
#                    Ej.: lenguaje (Java, C#, Python) -> 100, 010, 001.
#
# LabelEncoder    -> variable OBJETIVO (y) con varias clases.
#                    Convierte cada etiqueta en un número: 0, 1, 2...
#                    Ej.: especie (setosa, versicolor, virginica) -> 0, 1, 2.
#
# LabelBinarizer  -> variable OBJETIVO (y).
#                    Si hay 2 clases -> 0 / 1 (Ej.: contrato SI/NO -> 1/0).
#                    Si hay más de 2 -> una columna por clase (one-hot para y).
# ------------------------------------------------------------------------------
```

```python
x_num= df[[  "experiencia", "pret_salarial"]] #OBETENMOS LAS COLUMNAS NUMERICAS
   # x_num
x_nominal=df[["lenguaje"]]
   #     x_nominal
x_ordinal = df[["rango"]]
#x_ordinal
```

```python
numeric_scaler= StandardScaler()
nominal_encoder= OneHotEncoder(sparse_output=False)
ordinal_encoder = LabelEncoder()
print(nominal_encoder)
```

```python
x_num_scaled = numeric_scaler.fit_transform(x_num) #NOMRALIZAMOS NUESTRAS VARIABLES NUMERICAS
```

```python
x_nominal_encoded = nominal_encoder.fit_transform(x_nominal) #NORMALIZAMOS LAS VARIABLES NOMINALES
```

```python
x_ordinal_encoded= ordinal_encoder.fit_transform(x_ordinal)  #NOMALIZAMOS LAS VARIABLES ORDINALES
```

```python
import numpy as np
```

```python
data = np.hstack([
    x_num_scaled,                        # ya es 2D
    x_nominal_encoded.reshape(-1, 1),    # 1D -> 2D (una columna)
    x_ordinal_encoded.reshape(-1, 1)     # 1D -> 2D (una columna)
])
data
```
