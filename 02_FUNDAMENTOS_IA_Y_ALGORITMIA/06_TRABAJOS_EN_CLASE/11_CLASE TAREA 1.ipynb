```python
from sklearn.pipeline import Pipeline
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

from sklearn.inspection import DecisionBoundaryDisplay

from sklearn.feature_selection import SelectKBest, f_classif   #calcular cuales son los datos que mantienen mayor eficiencia para el modelo, si la intension es 2 mejores caracterisits es Fregression 
from sklearn.datasets import load_breast_cancer 
```

```python
data= load_iris()  # Cargamos el conjunto de datos de cáncer de mama
x= data.data  # Obtenemos las características (features)
y= data.target  # Obtenemos las etiquetas (labels)
```

```python
selector = SelectKBest(k=2, score_func=f_classif)  # Creamos un selector de características que seleccionará las 2 mejores características según la prueba
selector.fit_transform(x, y)  # Ajustamos el selector a los datos y transformamos las características
```

```python
indices= selector.get_support(indices=True)  # Obtenemos los índices de las características seleccionadas
print("Indices de las características seleccionadas para iris:", indices)  # Imprimimos los índices de las características seleccionadas
```

APLICAMOS LAS DOS MEJORES CARACTERISITCAS PARA EL MODELOS DE IRIS

```python
x, y = load_iris(return_X_y=True) #Obtenemos datos y etiquetas
x= x[:,[2,3]] #Seleccionamos solo las dos primeras características
print(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=42, stratify=y) #Dividimos en entrenamiento y prueba
```

```python
model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression())
])
model.fit(x_train, y_train)  # Entrenamos el modelo
```

```python
x, y = load_iris(return_X_y=True) #Obtenemos datos y etiquetas
x= x[:,[2,3]] #Seleccionamos solo las dos primeras características
#x

x, y = load_iris(return_X_y=True) #Obtenemos datos y etiquetas
load_iris().feature_names [1]
```

```python
DecisionBoundaryDisplay.from_estimator(model, x_train, response_method="predict", alpha=0.3, cmap="coolwarm")   #se genera el gráfico de la frontera de decisión del modelo entrenado, usando los datos de entrenamiento (x_train) y el método de respuesta "predict" para predecir las clases. El parámetro alpha controla la transparencia del área coloreada y cmap define el mapa de colores utilizado.         
plt.scatter(x_test[:, 0], x_test[:, 1], c=y_test, edgecolor="k", cmap="coolwarm") #scatter plot de los datos de prueba (pedimos d ela columna 0 y 1 de x_test, que son las dos características que seleccionamos)
```
