## 0. Preparar herramientas

```python
import numpy as np   # numpy es la libreria que hace todas las operaciones de algebra lineal por nosotros
```

## 1. Vectores

Un vector es simplemente una lista de números que representa una dirección o una cantidad.

```python
v1 = np.array([2, 3])   # creamos el vector 1: representa, por ejemplo, 2 unidades en x y 3 en y
v2 = np.array([1, 5])   # creamos el vector 2

suma = v1 + v2                 # sumamos los vectores: se suma posicion por posicion (2+1, 3+5)
print('Suma:', suma)           # mostramos el resultado

resta = v1 - v2                # restamos los vectores: igual, posicion por posicion
print('Resta:', resta)         # mostramos el resultado

escalar = v1 * 3               # multiplicamos el vector por un numero (escalar): agranda o achica el vector
print('v1 x 3:', escalar)      # mostramos el resultado

producto_punto = np.dot(v1, v2)          # producto punto: multiplica posicion por posicion y suma todo
print('Producto punto:', producto_punto) # nos dice que tan 'alineados' estan los dos vectores

norma = np.linalg.norm(v1)      # la norma es el 'largo' o magnitud del vector (como su tamano)
print('Norma de v1:', norma)    # mostramos el largo de v1
```

## 2. Matrices

Una matriz es una tabla de números (filas y columnas), como una mini hoja de cálculo.

```python
A = np.array([[1, 2], [3, 4]])   # creamos la matriz A, de 2 filas y 2 columnas
B = np.array([[5, 6], [7, 8]])   # creamos la matriz B, del mismo tamano

print('A + B:')          # sumamos las matrices
print(A + B)              # se suma numero por numero, en la misma posicion

print('A por 2:')         # multiplicamos la matriz por un numero (escalar)
print(A * 2)               # cada numero de la matriz se multiplica por 2

print('A multiplicada por B (matricial):')   # multiplicacion real entre matrices, no es igual a A*B
print(A @ B)                                  # el simbolo @ hace la multiplicacion matricial correcta

print('Transpuesta de A:')     # la transpuesta voltea filas por columnas
print(A.T)                      # A.T intercambia filas y columnas

print('Forma (shape) de A:', A.shape)   # shape nos dice cuantas filas y columnas tiene

identidad = np.eye(2)                # np.eye crea una matriz identidad (unos en la diagonal, ceros alrededor)
print('Matriz identidad:')            # es como el numero '1' pero para matrices
print(identidad)
```

## 3. Determinante e inversa

El determinante es un número que dice si una matriz se puede "deshacer" (invertir) o no.

```python
det_A = np.linalg.det(A)          # calculamos el determinante de A
print('Determinante de A:', det_A)  # si el determinante es 0, la matriz NO tiene inversa

if det_A != 0:                          # solo se puede calcular la inversa si el determinante no es 0
    inversa_A = np.linalg.inv(A)        # calculamos la matriz inversa de A
    print('Inversa de A:')
    print(inversa_A)

    comprobacion = A @ inversa_A        # una matriz multiplicada por su inversa debe dar la identidad
    print('A por su inversa (debe ser identidad):')
    print(np.round(comprobacion, 2))    # redondeamos para evitar numeros como 0.9999999
else:
    print('A no tiene inversa (determinante = 0)')
```

## 4. Sistemas de ecuaciones lineales

Ejemplo: resolver `2x + y = 5` y `x - y = 1` al mismo tiempo.

```python
# Las ecuaciones son:  2x + y = 5   y   x - y = 1

coeficientes = np.array([[2, 1], [1, -1]])   # los numeros que acompanan a x e y en cada ecuacion
resultados = np.array([5, 1])                 # los numeros del lado derecho del '='

solucion = np.linalg.solve(coeficientes, resultados)   # resuelve el sistema y encuentra x e y

print('x =', solucion[0])   # mostramos el valor encontrado de x
print('y =', solucion[1])   # mostramos el valor encontrado de y
```

## 5. Valores y vectores propios (eigenvalues/eigenvectors)

Son números y direcciones especiales de una matriz: muestran cómo 'estira' o 'gira' el espacio. Se usan mucho en IA para reducir dimensiones (por ejemplo, en PCA).

```python
valores, vectores = np.linalg.eig(A)   # calculamos los valores propios y vectores propios de A

print('Valores propios:', valores)      # numeros que dicen 'cuanto estira' la matriz en cada direccion especial
print('Vectores propios:')              # las direcciones especiales que la matriz solo estira (no gira)
print(vectores)
```

## Resumen simple

- **Vectores** → listas de números con dirección y tamaño (suma, resta, producto punto, norma).
- **Matrices** → tablas de números (suma, multiplicación `@`, transpuesta `.T`).
- **Determinante** → dice si una matriz se puede invertir (si es 0, no se puede).
- **Inversa** → la matriz que 'deshace' a otra (como dividir, pero para matrices).
- **Sistemas de ecuaciones** → `np.linalg.solve()` encuentra las incógnitas de varias ecuaciones a la vez.
- **Valores/vectores propios** → direcciones especiales de una matriz, muy usadas en IA (ej. PCA).
