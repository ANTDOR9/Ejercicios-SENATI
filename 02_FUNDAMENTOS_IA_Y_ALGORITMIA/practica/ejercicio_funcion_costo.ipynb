# Ejercicio extra: Función de costo (el "tazón" frío/caliente)

Esto recrea la gráfica que viste en clase: una superficie en 3D donde el color muestra qué tan grande es el error del modelo según sus parámetros.

**Idea clave:** azul/frío = poco error (bien) — rojo/caliente = mucho error (mal). El objetivo de entrenar un modelo es "caminar" desde el rojo hasta el azul.

## Paso 1: preparar las herramientas

```python
import numpy as np                       # para hacer los calculos numericos
import matplotlib.pyplot as plt          # para dibujar graficas
from mpl_toolkits.mplot3d import Axes3D  # para poder dibujar en 3 dimensiones (x, y, z)
```

## Paso 2: crear los parámetros de prueba (w y b)

Imaginemos que `w` y `b` son los dos parámetros que el modelo va probando, como en el perceptrón.

```python
w = np.linspace(-2, 2, 40)   # generamos 40 valores de w, entre -2 y 2 (el eje 'x' del dibujo)
b = np.linspace(-2, 2, 40)   # generamos 40 valores de b, entre -2 y 2 (el eje 'y' del dibujo)

W, B = np.meshgrid(w, b)   # combinamos w y b en una cuadricula: probamos TODAS las combinaciones posibles
```

## Paso 3: calcular el error para cada combinación

Usamos una función de error simple: mientras más lejos esté `w` y `b` del punto ideal (0, 0), más grande es el error.

```python
# Error = w al cuadrado + b al cuadrado
# (mientras mas grande w o b, mas grande el error; en (0,0) el error es 0, el minimo posible)
Error = W**2 + B**2   # calculamos el error para cada combinacion de w y b al mismo tiempo

print('Error minimo:', Error.min())   # el punto mas 'frio' (fondo del tazon)
print('Error maximo:', Error.max())   # el punto mas 'caliente' (bordes del tazon)
```

## Paso 4: dibujar el tazón (igual que en la diapositiva)

```python
fig = plt.figure(figsize=(8, 6))                 # creamos el lienzo donde vamos a dibujar
ax = fig.add_subplot(111, projection='3d')       # le decimos que el dibujo sera en 3D

# dibujamos la superficie: cmap='jet' es la paleta de colores frio(azul)-caliente(rojo)
superficie = ax.plot_surface(W, B, Error, cmap='jet', edgecolor='k', linewidth=0.2)

ax.set_xlabel('w (peso)')       # nombre del eje x
ax.set_ylabel('b (sesgo)')      # nombre del eje y
ax.set_zlabel('Error')          # nombre del eje z (altura)
ax.set_title('Funcion de costo: azul = poco error, rojo = mucho error')  # titulo de la grafica

fig.colorbar(superficie, shrink=0.5, label='Error (frio=bajo, caliente=alto)')  # barra de colores a un lado

plt.show()   # mostramos la grafica en pantalla
```

## Notas simples (frío vs. caliente)

- **Zona azul (fondo del tazón)** → el error es muy bajo → el modelo está prediciendo bien → esto es lo que se busca.
- **Zona roja (bordes, arriba)** → el error es muy alto → el modelo está prediciendo mal → hay que corregir `w` y `b`.
- **Entrenar un modelo = bajar del rojo hacia el azul**, ajustando `w` y `b` poco a poco. A eso se le llama *descenso de gradiente*.
- El punto exacto más azul de todos (aquí sería `w=0, b=0`) es el **mínimo global**: la mejor combinación posible de parámetros.
