## 1. Perceptrón + Sigmoid

Un perceptrón toma datos, los pesa, y decide algo (como aprobar o no).

```python
import numpy as np  # numpy nos da funciones matematicas listas para usar (como e^x)

# --- Ejercicio 1: aprobar o no aprobar segun el perceptron ---

x = 4       # el dato de entrada (por ejemplo, horas estudiadas)
w = 0.5     # el peso: que tanto importa ese dato
b = -3      # el sesgo (bias): un ajuste fijo que sube o baja el resultado

z = x * w + b       # formula del perceptron: multiplica el dato por su peso y suma el sesgo
print('z =', z)     # mostramos el resultado para verlo

sigmoid = 1 / (1 + np.exp(-z))   # sigmoid convierte cualquier numero en un valor entre 0 y 1 (como una probabilidad)
print('sigmoid(z) =', sigmoid)  # mostramos la probabilidad calculada

if sigmoid > 0.5:                     # si la probabilidad pasa la mitad (umbral)...
    print('Resultado: Aprueba')       # ...entonces decimos que aprueba
else:                                 # si no llega a la mitad...
    print('Resultado: No aprueba')    # ...entonces no aprueba
```

## 2. Softmax

Cuando hay MÁS de 2 opciones (no solo aprobar/no aprobar), usamos softmax para repartir probabilidades entre todas.

```python
# --- Ejercicio 2: elegir la clase mas probable entre varias opciones ---

puntajes = np.array([2, 0, 1])   # 3 puntajes crudos, uno por cada clase (ej: gato, perro, ave)

exponenciales = np.exp(puntajes)          # convertimos cada puntaje en un numero positivo con e^puntaje
print('exponenciales:', exponenciales)    # mostramos esos numeros intermedios

suma_total = np.sum(exponenciales)        # sumamos todos los numeros exponenciales
print('suma total:', suma_total)          # mostramos la suma (nos sirve para repartir el 100%)

probabilidades = exponenciales / suma_total   # cada numero se divide entre la suma -> queda como probabilidad
print('probabilidades:', probabilidades)      # mostramos las probabilidades finales (deben sumar 1)

clase_elegida = np.argmax(probabilidades)     # argmax busca la POSICION del valor mas alto
print('clase con mayor probabilidad: indice', clase_elegida)  # esa posicion es la clase ganadora
```

## 3. KNN — aprendizaje supervisado

Supervisado = le damos ejemplos YA etiquetados (ya sabemos la respuesta correcta) y el modelo aprende de ellos.

```python
from sklearn.neighbors import KNeighborsClassifier  # traemos el algoritmo KNN ya hecho por la libreria

# --- Ejercicio 3: clasificar una fruta nueva mirando a sus 'vecinos' mas parecidos ---

datos = [[150, 7], [170, 7.5], [140, 6.5], [130, 6], [180, 8], [120, 6]]  # cada fila es [peso en gramos, tamano en cm]
etiquetas = [0, 0, 0, 1, 0, 1]   # la respuesta correcta de cada fila: 0 = manzana, 1 = naranja

modelo = KNeighborsClassifier(n_neighbors=3)  # creamos el modelo: mirara a los 3 vecinos mas cercanos
modelo.fit(datos, etiquetas)                  # fit = 'entrenar': el modelo memoriza los ejemplos y sus respuestas

fruta_nueva = [[145, 6.8]]                    # una fruta nueva de la que NO sabemos si es manzana o naranja
prediccion = modelo.predict(fruta_nueva)      # predict = le pedimos al modelo que adivine, comparando con sus vecinos

print('Prediccion (0=manzana, 1=naranja):', prediccion)  # mostramos que adivino el modelo
```

## 4. K-Means — aprendizaje no supervisado

No supervisado = NO le damos respuestas correctas. El modelo agrupa los datos solo, segun que tan parecidos son.

```python
from sklearn.cluster import KMeans  # traemos el algoritmo K-Means ya hecho por la libreria

# --- Ejercicio 4: agrupar las mismas frutas, pero SIN decirle cual es cual ---

datos = [[150, 7], [170, 7.5], [140, 6.5], [130, 6], [180, 8], [120, 6]]  # los mismos datos de antes, sin etiquetas

modelo_kmeans = KMeans(n_clusters=2, random_state=0, n_init=10)  # le pedimos que forme 2 grupos (clusters)
modelo_kmeans.fit(datos)                                          # fit = el modelo busca solo los patrones y agrupa

print('Grupo asignado a cada fruta:', modelo_kmeans.labels_)         # a que grupo (0 o 1) quedo cada fruta
print('Centro de cada grupo:', modelo_kmeans.cluster_centers_)       # el 'promedio' de cada grupo formado

# Comparacion: revisa si modelo_kmeans.labels_ se parece a la lista 'etiquetas' del ejercicio de KNN.
# Si se parecen, quiere decir que el agrupamiento automatico coincidio con la clasificacion real.
```

## 5. Resumen simple

- **Perceptrón + Sigmoid** → toma datos, pesa su importancia, decide entre 2 opciones (si/no).
- **Softmax** → igual que sigmoid pero para MÁS de 2 opciones, reparte probabilidades entre todas.
- **KNN** → aprendizaje **supervisado**: ya conoce las respuestas correctas y compara un dato nuevo con sus vecinos.
- **K-Means** → aprendizaje **no supervisado**: no conoce respuestas, agrupa datos parecidos por su cuenta.
