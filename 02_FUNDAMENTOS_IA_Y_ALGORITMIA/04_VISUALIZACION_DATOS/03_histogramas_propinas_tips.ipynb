```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips') #Seaborn permite obtener el dataset como DataFrame
```

```python
tips.head()
```

```python
plt.hist(tips['tip'], bins=6, edgecolor='black', color='skyblue')
```

```python
hombres= tips[tips['sex']=='Male']
hombres
```

```python
plt.hist(hombres['tip'], bins=6, edgecolor='black', color='skyblue')
```

```python
# --- HISTOGRAMA DE PROPINAS (MUJERES) ---
mujeres = tips[tips['sex'] == 'Female']
mujeres

plt.hist(mujeres['tip'], bins=6, edgecolor='black', color='pink')

plt.title('Histograma de propinas - Mujeres')
plt.xlabel('Propina')
plt.ylabel('Frecuencia')

plt.show()
```

```python
# --- REPORTE ESTADÍSTICO 1: SUMA POR SEXO ---
suma_por_sexo = tips.groupby('sex')['tip'].sum()

print('Suma de propinas por sexo:')
print(suma_por_sexo)
```
