# Taller 5 — Tipos de gráficos básicos con Matplotlib

**Objetivo de esta práctica:** conocer los cinco gráficos básicos de Matplotlib usando listas de datos sencillas: barras, dispersión, líneas, torta e histograma.

Fue el primer ejercicio de visualización: antes de usar datos reales, se practica qué muestra cada gráfico y qué parámetros acepta.

## 1. Importar librería

```python
import matplotlib.pyplot as plt
```

## 2. Gráfico de barras

Sirve para **comparar categorías**. Se puede dar un color distinto a cada barra, y agregar título, nombres de ejes y cuadrícula.

```python
plt.bar(["A", "B", "C"], [15, 20, 10], color=["r", "black", "yellow"])
plt.title("Gráfico de barras 1")
plt.xlabel("Categorías")
plt.ylabel("Conteo")
plt.grid(-1)
```

```python
plt.bar(["a", "b", "c"], [48, 85, 60])
plt.title("Votos por serie")
```

## 3. Gráfico de dispersión

Muestra la **relación entre dos variables numéricas**: cada punto es un par (x, y). Con `marker` se cambia la forma del punto.

```python
plt.scatter([1, 2, 3], [2, 4, 6], marker="")
plt.title("Gráfico de dispersión")
```

```python
plt.scatter([5, 7, 8, 5, 6, 9], [110, 120, 130, 105, 115, 135], marker="v")
```

## 4. Gráfico de líneas

Une los puntos en orden; se usa para ver **tendencias**, por ejemplo cómo cambia un valor en el tiempo.

```python
plt.plot([1, 2, 3], [2, 4, 6])
plt.title("Gráfico de líneas")
```

## 5. Gráfico de torta

Muestra **proporciones** de un total. `autopct` escribe el porcentaje de cada parte.

```python
plt.pie([100, 100], labels=["A", "B"], autopct="%1.1f%%")
```

## 6. Histograma

Muestra la **distribución** de un conjunto de valores: cuántos caen en cada rango. `bins` define cuántos rangos (barras) se usan.

```python
plt.hist([20, 10, 18, 10, 21, 19, 20, 21, 22, 23, 24, 25], bins=2, edgecolor="green")
```

```python
plt.hist([12, 15, 14, 19, 11, 15, 16, 14, 15, 18, 20, 11, 14, 15, 16], bins=3)
plt.title("Histograma con 3 rangos")
```

## 🗒️ Conclusión

| Gráfico | Función | Se usa para |
|---|---|---|
| Barras | `plt.bar` | Comparar categorías |
| Dispersión | `plt.scatter` | Ver la relación entre dos variables |
| Líneas | `plt.plot` | Ver tendencias |
| Torta | `plt.pie` | Mostrar proporciones |
| Histograma | `plt.hist` | Ver la distribución de los datos |
