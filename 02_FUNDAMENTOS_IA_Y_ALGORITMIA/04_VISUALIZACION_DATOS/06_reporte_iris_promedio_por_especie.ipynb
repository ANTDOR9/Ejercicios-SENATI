# Taller 6 — Reporte de promedios por especie con el dataset Iris

**Objetivo de esta práctica:** aplicar el flujo de reportes (`groupby` → `.index` / `.values` → gráfico) a un dataset real: Iris. Se calcula el promedio de cada medida por especie y se comparan dos medidas en un mismo gráfico de barras.

## 1. Importar librerías

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris      # Dataset Iris incluido en scikit-learn
```

## 2. Cargar el dataset

```python
iris = load_iris()
```

## 3. Convertir a DataFrame y agregar la especie

`iris.data` trae solo las medidas. Se agrega la columna `species` con el nombre de la especie, usando `iris.target` (0, 1, 2) para buscar el nombre en `iris.target_names`.

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["species"] = iris.target_names[iris.target]
df.head()
```

## 4. Reporte: promedio de cada medida por especie

```python
promedio = df.groupby("species").mean()
promedio
```

## 5. Separar etiquetas (`.index`) y valores (`.values`)

```python
print(promedio.values)
print(promedio.index)
```

## 6. Gráfico: largo del pétalo vs. largo del sépalo por especie

Se dibujan dos series de barras sobre las mismas categorías; `alpha=0.5` las hace semitransparentes para que se vean ambas.

```python
plt.bar(promedio["petal length (cm)"].index, promedio["petal length (cm)"].values, alpha=0.5)
plt.bar(promedio["sepal length (cm)"].index, promedio["sepal length (cm)"].values, alpha=0.5)
```

## 🗒️ Conclusión

El mismo flujo de los reportes de ventas funciona con cualquier dataset: agrupar, separar etiquetas y valores, y graficar. En Iris se ve que *virginica* tiene en promedio los pétalos y sépalos más largos, y *setosa* los más cortos.
