```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

ventas = pd.DataFrame({
    'vendor': [
        'Ana', 'Ana','Bruno', 'Bruno','Carlos', 'Carlos','Diego',
        'Diego','Elena', 'Elena'
        ],'ventas': [200, 500,400, 600,300, 500,400, 800, 700, 900
  ]})
```

```python
# --- Funciones de varias filas ---
# Venta máxima
venta_max = ventas['ventas'].max()
print("Venta máxima:", venta_max)
# Conteo de ventas
conteo_ventas = ventas['ventas'].count()
print("Conteo de ventas:", conteo_ventas)
# Media de ventas
media_ventas = ventas['ventas'].mean()
print("Media de ventas:", media_ventas)
# Suma de ventas
suma_ventas = ventas['ventas'].sum()
print("Suma de ventas:", suma_ventas)
```

```python
# Trabajando con Bruno
# Todo de Bruno
print(ventas[ventas['vendor'] == 'Bruno'])
# Suma de las ventas de Bruno
print(ventas[ventas['vendor'] == 'Bruno']['ventas'].sum())
# Venta máxima de Bruno
print(ventas[ventas['vendor'] == 'Bruno']['ventas'].max())
```

```python
# Todo de Diego
print(ventas[ventas['vendor'] == 'Diego'])

# Suma de las ventas de Diego
print(ventas[ventas['vendor'] == 'Diego']['ventas'].sum())

# Todo de Bruno
print(ventas[ventas['vendor'] == 'Bruno'])

# Suma de las ventas de Bruno
print(ventas[ventas['vendor'] == 'Bruno']['ventas'].sum())

# --- Trabajando con Diego y Bruno ---

# Venta máxima de Diego
print("Venta máxima de Diego:",
      ventas[ventas['vendor'] == 'Diego']['ventas'].max())

# Promedio de todas las ventas de Diego
print("Promedio de Diego:",
      ventas[ventas['vendor'] == 'Diego']['ventas'].mean())

# Venta máxima de Bruno
print("Venta máxima de Bruno:",
      ventas[ventas['vendor'] == 'Bruno']['ventas'].max())

# Promedio de todas las ventas de Bruno
print("Promedio de Bruno:",
      ventas[ventas['vendor'] == 'Bruno']['ventas'].mean())
```

```python
# --- TRABAJANDO CON TODOS LOS VENDEDORES ---

# SUMA TOTAL POR VENDEDOR
suma_por_vendedor = ventas.groupby('vendor')['ventas'].sum()
print(suma_por_vendedor)

# PROMEDIO DE VENTAS POR VENDEDOR
promedio_por_vendedor = ventas.groupby('vendor')['ventas'].mean()
print(promedio_por_vendedor)
```

```python
# --- COMBINANDO GRÁFICOS ---

print('Suma por vendedor:')
print(suma_por_vendedor.index)
print(suma_por_vendedor.values)

print('Promedio por vendedor:')
print(promedio_por_vendedor.index)
print(promedio_por_vendedor.values)

plt.bar(suma_por_vendedor.index, suma_por_vendedor.values, label='Suma')
plt.bar(promedio_por_vendedor.index, promedio_por_vendedor.values, label='Promedio')

plt.title('Suma y promedio por vendedor')
plt.xlabel('Vendedor')
plt.ylabel('Ventas')
plt.legend()
plt.xticks(rotation=45)

plt.show()
```
